# Student Performance Analytics System

## Python with AI Mini Project

### Objective
To build a Python-based Student Performance Analytics System that reads,
processes, and analyzes student performance data using Pandas and NumPy.

### Technologies Used
- Python
- Pandas
- NumPy
- Google Colab

### Dataset
The dataset contains 30 student records with:
- Student ID
- Name
- Department
- Python marks
- Mathematics marks
- DBMS marks
- AI marks
- Attendance

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("students.csv")

print(df)
```

```python
# Display basic information about the dataset

print("Number of students:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nBasic statistics:")
print(df.describe())
```

```python
# Define the subjects
subjects = ["Python", "Mathematics", "DBMS", "AI"]

# Calculate total marks using NumPy
df["Total_Marks"] = np.sum(df[subjects].values, axis=1)

# Calculate average marks
df["Average_Marks"] = df["Total_Marks"] / len(subjects)

# Display the result
print(df[["Student_ID", "Name", "Total_Marks", "Average_Marks"]])
```

```python
# Function to assign grades

def assign_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


# Apply the function to every student
df["Grade"] = df["Average_Marks"].apply(assign_grade)

# Display results
print(df[["Student_ID", "Name", "Average_Marks", "Grade"]])
```

```python
# Function to determine pass/fail status

def check_result(average, attendance):
    if average >= 40 and attendance >= 75:
        return "Pass"
    else:
        return "Fail"


# Apply the function to the dataset
df["Result"] = df.apply(
    lambda row: check_result(row["Average_Marks"], row["Attendance"]),
    axis=1
)

# Display results
print(df[["Student_ID", "Name", "Average_Marks", "Attendance", "Result"]])
```

```python
# Attendance vs Average Marks

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Attendance"],
    df["Average_Marks"]
)

plt.title("Attendance vs Average Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")
plt.xlim(60, 100)
plt.ylim(0, 100)

plt.show()
```

```python
# Find the student with the highest average
highest_student = df.loc[df["Average_Marks"].idxmax()]

# Find the student with the lowest average
lowest_student = df.loc[df["Average_Marks"].idxmin()]

# Calculate overall class average
class_average = np.mean(df["Average_Marks"])

print("===== OVERALL PERFORMANCE =====")

print("\nHighest Performing Student:")
print("Name:", highest_student["Name"])
print("Student ID:", highest_student["Student_ID"])
print("Average Marks:", highest_student["Average_Marks"])

print("\nLowest Performing Student:")
print("Name:", lowest_student["Name"])
print("Student ID:", lowest_student["Student_ID"])
print("Average Marks:", lowest_student["Average_Marks"])

print("\nOverall Class Average:", round(class_average, 2))
```

```python
# Calculate the average marks for each subject

subject_averages = df[subjects].mean()

print("===== SUBJECT-WISE PERFORMANCE =====")

for subject, average in subject_averages.items():
    print(f"{subject}: {average:.2f}")

# Find the subject with the highest average
best_subject = subject_averages.idxmax()
best_subject_average = subject_averages.max()

print("\nBest Performing Subject:")
print("Subject:", best_subject)
print("Average Marks:", round(best_subject_average, 2))
```

```python
# Find the top 5 performing students

top_students = df.sort_values(
    by="Average_Marks",
    ascending=False
).head(5)

print("===== TOP 5 PERFORMING STUDENTS =====")

print(
    top_students[
        ["Student_ID", "Name", "Average_Marks", "Grade", "Result"]
    ].to_string(index=False)
)
```

```python
# Calculate pass/fail statistics

pass_count = (df["Result"] == "Pass").sum()
fail_count = (df["Result"] == "Fail").sum()

total_students = len(df)

pass_percentage = (pass_count / total_students) * 100
fail_percentage = (fail_count / total_students) * 100

print("===== PASS / FAIL STATISTICS =====")

print("Total Students:", total_students)
print("Students Passed:", pass_count)
print("Students Failed:", fail_count)
print("Pass Percentage:", round(pass_percentage, 2), "%")
print("Fail Percentage:", round(fail_percentage, 2), "%")
```
